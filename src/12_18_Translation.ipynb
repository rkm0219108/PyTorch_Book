# 翻譯(Translation)

## 載入相關套件

```python
# 載入相關套件
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer, pipeline
```

## 載入模型

```python
translation_model = AutoModelForSeq2SeqLM.from_pretrained("t5-base")
translation_tokenizer = AutoTokenizer.from_pretrained("t5-base")

def translator(text: str, max_length=40, **kwargs):
    inputs = translation_tokenizer("translate English to German: " + text, return_tensors="pt")
    outputs = translation_model.generate(inputs["input_ids"], max_length=max_length, **kwargs)
    return [{"translation_text": translation_tokenizer.decode(outputs[0], skip_special_tokens=True)}]
```

## 測試

```python
text = "Hugging Face is a technology company based in New York and Paris"
print(translator(text, max_length=40))
```

## 結合Tokenizer

```python
model = AutoModelForSeq2SeqLM.from_pretrained("t5-base")

# 結合分詞器(Tokenizer)
tokenizer = AutoTokenizer.from_pretrained("t5-base")
text = "translate English to German: Hugging Face is a " + "technology company based in New York and Paris"
inputs = tokenizer(text, return_tensors="pt")
outputs = model.generate(inputs["input_ids"], max_length=40, num_beams=4, early_stopping=True)

print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
