# [優化器(Optimizers)](https://pytorch.org/docs/stable/optim.html#algorithms)

```python
# 載入套件
import torch
from torch import nn, optim
```

## 隨機梯度下降法(Stochastic Gradient Decent, SGD)

```python
# 建立模型
model = nn.Sequential(
    nn.Flatten(),
    nn.Linear(28 * 28, 256),
    nn.Dropout(0.2),
    nn.Linear(256, 10),
)

criterion = nn.CrossEntropyLoss()

# 隨機梯度下降法(SGD)
optimizer = optim.SGD(model.parameters(), lr=0.1, momentum=0.9)

optimizer.zero_grad()
input = torch.randn(3, 28 * 28, requires_grad=True)
target = torch.empty(3, dtype=torch.long).random_(5)  # 目標值
loss = criterion.forward(model(input), target)
loss.backward()
optimizer.step()
```

## Adam(Adaptive Moment Estimation)

```python
# 隨機梯度下降法(SGD)
optimizer = optim.Adam(model.parameters(), lr=0.1)

optimizer.zero_grad()
input = torch.randn(3, 28 * 28, requires_grad=True)
target = torch.empty(3, dtype=torch.long).random_(5)  # 目標值
loss = criterion.forward(model(input), target)
loss.backward()
optimizer.step()
```
