# 激勵函數(Activation Functions)

```python
# 載入套件
import torch
from torch import nn
```

## ReLU

```python
m = nn.ReLU()
input = torch.tensor([5, 2, 0, -10])
output = m(input)
output
```

## LeakyReLU

```python
m = nn.LeakyReLU()
input = torch.tensor([5, 2, 0, -10, -100], dtype=torch.float)
output = m(input)
output
```

## Sigmoid

```python
m = nn.Sigmoid()
input = torch.tensor([5, 2, 0, -10, -100], dtype=torch.float)
output = m(input)
output
```

## Tanh

```python
m = nn.Tanh()
input = torch.tensor([5, 2, 0, -10, -100], dtype=torch.float)
output = m(input)
output
```

## Softmax

```python
m = nn.Softmax(dim=1)
input = torch.tensor([[1.0, 2.0, 3.0, 4.0]], dtype=torch.float)
output = m(input)
output
```
